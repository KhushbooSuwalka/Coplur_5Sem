```python
# PIVOTING A DATAFRAME IN PANDAS
import pandas as pd

df= pd.DataFrame({
    'Col1': range(12),
    'Col2': ['A']*3 +['B']*3 + ['C']*3 + ['D']*3,
    'Date':pd.to_datetime(['2023-01-01', '2023-01-02', '2023-01-03']*4)
    # 'Date': pd.date_range('2023-01-01', periods=12, freq='D')
})

print("Original DataFrame:")
print(df)

pivoted= df.pivot(index='Date', columns='Col2', values='Col1')
print("\nPivoted DataFrame:")
print(pivoted)
```

```python
data={
    'Course':['CS','Civil','DS','CS','Civil'],
    'Year':[1,2,1,2,1],
    'Student':[100,200,300,400,500]
}
df=pd.DataFrame(data)
print("Original DataFrame :")
print(df)

df.loc[5]=['MC',2,600]
print("\nDataFrame after adding a new row:")
print(df)

df['Marks']=[85,90,95,80,75,88]  #The values to the column should be of the same length as the number of rows in the DataFrame,otherwise error is generated
print("\nDataFrame after adding a new column:")
print(df)

df['Score']=df['Student'] #Copy the values of the 'Student' column to the new 'Score' column
print("\nDataFrame after copying 'Student' column to 'Score' column:")
print(df)
```

```python
df['Marks']=df['Student'][:3]
print("\nDataFrame after modifying 'Marks' column with values from 'Student' column:")
print(df)

pivoted=df.pivot_table(
    index="Course",
    values="Student",
    columns="Year",
    aggfunc="sum"
)
print("\nPivoted table:")
print(pivoted)
```

```python
# Handling Missing Values
import pandas as pd
import numpy as np

data={
    'Col1':[1,np.nan,np.nan,4],
    'Col2':[1,pd.NA,pd.NA,2]
}
df=pd.DataFrame(data)
print("DataFrame:\n",df)

# Fill missing values with a specific value (e.g., 0)
filled_df=df.fillna(0)
print("\nDataFrame after filling missing values:\n",filled_df)

# Fill missing values with "-"
filled_df=df.fillna("-")
print("\nDataFrame after filling missing values with '-':\n",filled_df)
```

```python
res=df.ffill()  # Fill missing values  with the previous value in the same column
print("Forward fill:\n",res) 

res=df.bfill()  # Fill missing values with the next value in the same column
print("Backward fill:\n",res)
```

```python
import pandas as pd
import numpy as np
data=[
    [1,2,3],
    [4,5,6],
    [7,8,9]
]
df=pd.DataFrame (data,index=['a','c','d'],columns=['col1','col2','col3'])
print(df)

df1=df.reindex(['a','b','c','d','e'])
print("\nAfter reindexing:\n",df1)

df2=df.reindex(['a','b','c','d','e','f'])
print("\nAfter reindexing with more labels:\n",df2)
```

```python
df1=df1.ffill(limit=1)
print("\nAfter forward filling with limit=1 of df1:\n",df1)

df2=df2.ffill(limit=1)  # Fill missing values with the previous value in the same column, but only for a maximum of 1 consecutive missing value
print("\nAfter forward filling with limit=1 of df2:\n",df2)
```

```python
# Aggregations with groupby()
data={
    'one': ["A","B","A","C","A","C","A"],
    'two': [5,60,7,2000,45,78,85]
}
df= pd.DataFrame(data)
print("DataFrame group by 'one' column and sum of 'two' column:")
gd= df.groupby('one')['two'].sum()
print(gd)

print("\nDataFrame group by 'one' column and mean of 'two' column:")
gd_mean= df.groupby('one')['two'].mean()
print(gd_mean)
```

```python
data={
    'one': ["A","B","A","C","A","C","A"],
    "SubCategory": ["X","Y","X","Z","Y","Z","X"],
    'two': [5,60,7,2000,45,78,85]
}
df= pd.DataFrame(data)
print("DataFrame group:") #DataFrame group by 'one' and 'SubCategory' columns and sum of 'two' column
gd_subcat= df.groupby(['one','SubCategory'])['two'].sum()
print(gd_subcat)
```

```python
res= df.sort_values(by='two')
print("\nDataFrame sorted by 'two' column:\n",res) #By default,ascending order is used

res= df.sort_values(by='two', ascending=False)
print("\nDataFrame sorted by 'two' column (descending):\n",res)
```

```python
res= df.sort_values(by=['one','two'])
print("\nDataFrame sorted by 'one' and then by 'two' columns:\n",res)

res= df.sort_values(by=['one','two'],ascending=[True,False]) # Sort by 'one' column in ascending order and then by 'two' column in descending order
print("\nDataFrame sorted by 'one' and then by 'two' columns:\n",res)
```
