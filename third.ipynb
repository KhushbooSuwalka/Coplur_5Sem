```python
#sets:
set1={"abc",1, "python","network",1,True, False,0} #false & 0 are considered same
set1.add("java")
set2=set([1,2,3,4,5])
print(set1, type(set1))
print(set2, type(set2))
print(set1.union(set2))
print(set1 | set2)

#initialising s1 and s2 so that wo pta kar sake ki ye set banega
s1=set()
s2=set()

for i in range(5):
    s1.add(i)
print(s1)

for i in range(3,10):
    s2.add(i)
print(s2)

#same methods for intersection of sets:
s3=s1.intersection(s2)
print(s3)
s4=s1 & s2
print(s4)
```

```python
#functions:

#positional arguments:(values are taken in order in an argument)
def function1(n1,n2):
    print("example of fn")
    print(n1*n2)
function1(10,20)

def function2(*city): #we have used *city and not just city bcoz we dont know how many cities will be passed as an argument from the positional argument
    print(city)
    print(city[0])
function2("jpr","ajmer","kota")

#keyword arguments:(no order is prescribed for the arguments to take values)
def fun1(city3,city2,city1):
    print(city2)
fun1(city1="mumbai", city3="ajmer", city2="kota")

# unknown positional: use *
# unknown keyword: use **

def fun2(city="jpr"):
    print(city)
fun2()
fun2("new delhi")

def fun3(fruit):
    for x in fruit:
        print(x)
fruits=["apple","banana", "cherry"]
fun3(fruits)

#two methods, same output
#M-1:
def dob(*,n1):
    print(n1*2)
dob(n1=5)
#M-2:
def dob(n1,/):
    print(n1*2)
dob(2)

def dob3(a,b,/,*,c,d):
    print(a+b+c+d)
dob3(1,2,c=3,d=4) #a,b:positional and c,d: keyword
#a and b ko normally values denge but c and d ke liye alag se specify karna pdega bcoz uss se pehle operators aa gye so computer samjh nhi paayega ki given value kiski hai


#doubt: why both if and else block are executing at the same time?
def rec(k):
    if k>0:
        print(k)
        k=k-1
        rec(k)
    else: 
         print("rec not called")
rec(5)
```
