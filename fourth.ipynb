```python
import csv

d=[['Name', 'Age', 'City'],
   ['Alice', 30, 'New York'],
   ['Bob', 25, 'Los Angeles'],
   ['Charlie', 35, 'Chicago']]

# write to csv file
with open('data.csv', 'w',newline='') as file:
    writer = csv.writer(file)
    # writer.writerows(d)
    for row in d:
        writer.writerow(row)
print("Data written to data.csv")

# read csv file
with open('data.csv', 'r') as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

```python
# exception handling
try:
    n1=int(input("Enter a divisor : "))
    result=35/n1
    print("Result : ",result)
except ZeroDivisionError:
    print("Error: Division by zero is not allowed.")    
except ValueError:
    print("Error: Invalid input. Please enter a valid integer.")
except Exception as e:
    print("An error occurred:", str(e)) 

#Index error
try:
    lst=[1,2,3]
    print(lst[5])
except IndexError:
    print("Error: Index out of range.")
```
