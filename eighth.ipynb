COPY AND VIEW OF ARRAY

```python
arr=np.array([1,2,3,4,5])
x=arr.copy()
arr[0]=50 #Changes the element at index 0 to 50 but it does not change the element at index 0 in x because x is a copy of arr and they are stored in different memory locations.
print(arr)
print(x)
```

```python
arr = np.array([1, 2, 3, 4, 5, 6, 7])
x=arr.view() # It creates a view of the original array. Changes made to the view will affect the original array and vice versa because they are stored in the same memory location.
arr[0]=50
print(arr)
print(x)
```

```python
a= np.array([
    [1,2,3],
    [4,5,6]
])
print(a)
a1=a.reshape(6) # It changes the shape of the array without changing its data. The total number of elements must remain the same.
print(a1)
a2=a.reshape(3,2) # It changes the shape of the array to 3 rows and 2 columns. The total number of elements must remain the same.
print(a2)
```
