```python
import sqlite3
conn = sqlite3.connect('example.db')
sql="""
CREATE TABLE emp(
    id INTEGER PRIMARY KEY,
    name VARCHAR(100) NOT NULL,
    age INTEGER,
    salary FLOAT(10, 2)
)
"""
conn.execute(sql)
conn.commit()
conn.close()
```

```python
conn = sqlite3.connect('example.db')
sql = "INSERT INTO emp(id,name, age, salary) VALUES(1,'Alice', 30, 50000.00)"
conn.execute(sql)
conn.commit()
```

```python
sql = "INSERT INTO emp VALUES(2,'Bob', 25, 45000.00),(3,'Charlie', 28, 48000.00), (4,'David', 35, 55000.00)"
conn.execute(sql)
conn.commit()
```

```python
sql="SELECT * FROM emp"
result = conn.execute(sql)
for row in result:
    print(row)
```

```python
sql = "DELETE FROM emp WHERE id=1"
conn.execute(sql)
conn.commit()

sql="SELECT * FROM emp ORDER BY salary DESC limit 2"
result = conn.execute(sql)
for row in result:
    print(row)
```

```python
sql = "UPDATE emp SET salary=salary*1.1 WHERE age>30"
conn.execute(sql)
conn.commit()
```
