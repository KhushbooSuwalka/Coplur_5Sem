```python
# ARITHMETIC OPERATIONS IN PANDAS DATAFRAME
import pandas as pd

df= pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
print(df)
print("\nAddition :\n",df+5)
print("\nSubtraction :\n",df-2)
print("\nMultiplication :\n",df*3)
print("\nDivision :\n",df/2)
print("\nModulus :\n",df%2)
print("\nExponentiation :\n",df**2)
```

```python
df1= pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
print("DataFrame 1 :\n",df1)

print("\nAddition :\n",df1['B']+5) # Adding 5 to each element in column 'B'
```

```python
df1= pd.DataFrame({'A': [1, 2, 3], 'B': [4, 5, 6]})
df2= pd.DataFrame({'A': [7, 8, 9], 'B': [10, 11, 12]})
print("\nDataFrame 1 :\n",df1)
print("\nDataFrame 2 :\n",df2)
print("\nAddition :\n",df1+df2) # Adding two DataFrames element-wise
```

```python
df1 = pd.DataFrame({'A': [1, 2, 3, 4], 'B': [4, 5, 6, 7]})
df2 = pd.DataFrame({'A': [7, 8, 9], 'B': [10, 11, 12]}, index=[1, 2, 3] )
print("DataFrame 1 :\n", df1)
print("\nDataFrame 2 :\n", df2)

print("\nAddition with different indices :\n", df1 + df2) # Adding two DataFrames with different indices
print("\nAddition with fill_value :\n", df1.add(df2, fill_value=0)) # Adding two DataFrames with fill_value for missing indices
print("\nSubraction with different indices :\n",df1-df2)
```

```python
one=pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Subject":["sub1","sub2","sub3"],
    "Marks":[85, 90, 95]
})
two=pd.DataFrame({
    "Name": ["Aliana", "Bahubali", "David"],
    "Subject":["sub4","sub5","sub3"],
    "Marks":[80, 85, 90]
},index=[1,2,3])
print("DataFrame 1 :\n", one)
print("\nDataFrame 2 :\n", two)

print("\nConcatenating along rows:")
res=pd.concat([one,two]) # Concatenating two DataFrames along rows (default axis=0)
print(res)
```

```python
one=pd.DataFrame({
    "Name": ["Alice", "Bob", "Charlie"],
    "Subject":["sub1","sub2","sub3"],
    "Marks":[85, 90, 95]
})
two=pd.DataFrame({
    "Name": ["Aliana", "Bahubali", "David"],
    "Subject":["sub4","sub5","sub3"],
    "Marks":[80, 85, 90]
},index=[1,2,3])

res=pd.concat([one,two],keys=['x','y']) # Concatenating two DataFrames with keys to create a hierarchical index
print(res)
```

```python
print("\nConcatenating along columns (axis=1) :\n")
res=pd.concat([one,two],axis=1) # Concatenating two DataFrames along columns (axis=1)
print(res)

print("\nConcatenating with ignore_index=True :\n")
res=pd.concat([one,two],ignore_index=True) # Concatenating two DataFrames and ignoring the original indices
print(res)
```

```python
one=pd.DataFrame({
    "id":[1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "Subject":["sub1","sub2","sub3"],
    "Marks":[85, 90, 95]
})
two=pd.DataFrame({
    "id":[1,2, 3],
    "Name": ["Aliana", "Bahubali", "David"],
    "Subject":["sub4","sub5","sub3"],
    "Marks":[80, 85, 90]
})
res=one.merge(two,on="id") # Merging two DataFrames based on a common column 'id'
print(res)
```

```python
one=pd.DataFrame({
    "id":[1, 2, 3],
    "Name": ["Alice", "Bob", "Charlie"],
    "Subject":["sub1","sub2","sub3"],
    "Marks":[85, 90, 95]
})
two=pd.DataFrame({
    "id":[2,1, 3,4],
    "Name": ["Aliana", "Bahubali", "David","Haya"],
    "Subject":["sub4","sub5","sub3","sub4"],
    "Marks":[80, 85, 90,95]
})
res=one.merge(two,on="id") # Merging two DataFrames based on a common column 'id'
print(res)
```

```python
res=two.merge(one,on="id")
print(res)
```

```python
res=one.merge(two,on=["id","Subject"]) 
# Merging two dataframes based on common id and subject both
print(res)
```

```python
res=one.merge(two,on="Subject",how="left")
print(res,"\n")
res=one.merge(two,on="Subject",how="right")
print(res)
```

```python
res=one.merge(two,on="Subject",how="inner")
print(res,"\n")
res=one.merge(two,on="Subject",how="outer")
print(res)
```

```python
res=one.join(two,lsuffix="left",rsuffix="right") 
print(res)
```
