```python
import pandas as pd

data=[10,20,30,40,50]
ps = pd.Series(data,index=["a","b","c","d","e"])
print (ps)
# print(ps[0])    
#If we do not have given index to the data, then use it  
print(ps["a"])
# Using above line because we have given index to data in this case
print(type(ps))
```

```python
data={
    "name":["Alice","Bob","Charlie"],
    "age":[25,30,35],
    "city":["New York","Los Angeles","Chicago"],
    "salary":[70000,80000,90000]
}
df=pd.DataFrame(data)
print(df)
```

```python
data={"a":[10,20,30],"b":[40,50,60]}
df=pd.DataFrame(data)
print(df)
print(df.loc[0])
print(df.loc[[0]])   # Data on index 0
print(df.loc[0:2])  # Data on index 0,1 and 2
```

```python
data={"a":[10,20,30],"b":[40,50,60]}
df=pd.DataFrame(data)
print(df)
df.index=["x","y","z"]
print(df)
df.columns=["col1","col2"]
print(df)
```

```python
df=pd.DataFrame({"a": [10, 20,30,40], "b": [30, 40,50,60], "c": [50, 60,70,80]})
print(df)
df.index=["x","y","z","w"]
df.columns=["col1","col2","col3"]
res=df.loc["x":"y","col1":"col2"]
print(res)
```

```python
df=pd.DataFrame({"a": [10, 20,30,40], "b": [30, 40,50,60], "c": [50, 60,70,80]})
print(df)

df.iloc[0:2,1] =[100,200] # Data on index 0 and 1 of column 1,i.e.,column b will be replaced by 100 and 200
print(df)
```

```python
df=pd.DataFrame([[10, 20,30,40],[30, 40,50,60],[50, 60,70,80]],columns=["col1","col2","col3","col4"])
print(df)

res=df.drop(0)
print(res)

res=df.drop([0,1])
print(res)

res=df.drop("col1",axis=1)
print(res)
```

```python
df=pd.DataFrame({"a": [10, 20,30,40], "b": [30, 40,50,60], "c": [50, 60,70,80]},index=["x","y","z","w"])
print(df)

result=df.drop(df.index[0:2])
print(result)
```

```python
df = pd.DataFrame({'A':[1,2,3,4,5],'B':[4,5,6,7,8],'C':[9,10,11,12,0]}, index = ['r1','r2','r3','r4','r5'])

print(df)
# drop row where columns C is 0

result = df[df["C"]!=0]
print(result,"\n")

print(result['C'])
```
